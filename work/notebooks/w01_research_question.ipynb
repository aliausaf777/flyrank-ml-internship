### Quick Look at the Data

The starter dataset contains **30,000 content pages** with **44 features**, providing a sufficient sample for exploring content refresh opportunities.

The calculated statistics from the starter dataset show that there are many pages with measurable search performance, making this dataset suitable for building a content opportunity scoring system and testing ranking approaches.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load the dataset
df = pd.read_csv("/content/flyrank-ml-internship/data/raw/content_refresh_anonymized.csv")

# Calculate a few useful statistics
total_pages = len(df)
avg_ctr = df["ctr"].mean()
high_impression_pages = (df["impressions_90d"] >= 500).sum()

print(f"Total pages: {total_pages:,}")
print(f"Average CTR: {avg_ctr:.2f}")
print(f"Pages with 500+ impressions: {high_impression_pages:,}")

print("\nSummary")
print(f"Dataset contains {total_pages:,} content pages.")
print(f"Average CTR: {avg_ctr:.2f}")
print(f"Pages with 500+ impressions: {high_impression_pages:,}")
```
